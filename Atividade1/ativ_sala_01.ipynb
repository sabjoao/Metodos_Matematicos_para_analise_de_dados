```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

Para as questões 1, 2 e 3 você usará o dataset movies, o qual contém informações de filmes do site IMDB no período de 1980 a 2020.

```python
df = pd.read_csv('movies.csv')
df.head()
```

### Questão 1
Obtenha a média, a mediana e o desvio padrão do score dos filmes cujo gênero é Drama.

```python
media = df[df['genre']=='Drama']['score'].mean()
mediana = df[df['genre']=='Drama']['score'].median()
desvio = df[df['genre']=='Drama']['score'].std()
print(media)
print(mediana)
print(desvio)
```

### Questão 2 
Gere um boxplot que mostre o valor do score dos seguintes gêneros: Animation, Comedy, Drama, Horror e Thriller. 

```python
generos = ['Animation', 'Comedy', 'Drama', 'Horror', 'Thriller']
dfGen = df[df['genre'].isin(generos)]
plt.Figure(figsize=(10,8))
sns.boxplot(data = dfGen, x = 'genre', y='score')
plt.title('Boxplot grouped by genre')
plt.ylabel('')
plt.show()
```

### Questão 4
O dataset "real_state_ny.csv" mostra dados de casas à venda no estado de Nova Iorque, nos EUA. Utilizando este dataset: \
a) Informe a média, a mediana e o desvio padrão do preço das casas de 3 quartos (bed = 3) da cidade de Albany (city = 'Albany') \
b) Gere um boxplot do preço do imóvel por número de quartos na cidade de Albany. Limite o número de quartos a no máximo 5.

```python
df3bed = df[(df['bed'] == 3) & (df['city'] == 'Albany')]
df3bed.head()

dfLim5 = df[(df['bed'] <= 5) & (df['city'] == 'Albany')]
plt.figure(figsize=(10,8))
sns.boxplot(data = dfLim5, x='bed', y='price')
plt.title('Preco por quarto')
plt.show()
```

### Questão 5
Plote um gráfico de pizza que mostre a porcentagem de imóveis por número de quartos para a cidade de Albany.

```python
dfAlb = df[df['city'] == 'Albany']
dfAlbBed = dfAlb['bed'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(dfAlbBed, autopct='%1.1f%%', labels= dfAlbBed.index)
plt.title('Porcentagem por número de quartos')
plt.show()
```

### Questão 6
Plote um gráfico de barras que mostre o preço médio das casas de acordo com o número de banheiros nas cidades de New York, Albany, Yonkers e Long Beach. Limite o número de banheiros a 4, ou seja, descarte os imóveis que tenham mais que 4 banheiros.

```python
lista = ['New York', 'Albany', 'Yonkers', 'Long Beach']
dfzz = df[(df['city'].isin(lista)) & (df['bath'] < 5)]
dfzz['media'] = dfzz['price'].mean()
plt.figure(figsize=(8,6))
sns.barplot(data = dfzz, x ='city', y ='price')
plt.show()

```
