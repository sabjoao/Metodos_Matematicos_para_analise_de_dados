```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
```

Questão 1 \
O peso médio de 500 estudantes do sexo masculino de uma universidade é 75,50 kg e o desvio padrão 7,50 kg. Admitindo-se que os pesos estão normalmente distribuídos e que as medidas foram realizadas com uma balança que registrou valores na escala de centésimos de quilogramas, determine quantos estudantes pesam: \
a) entre 60 e 77,5 kg \
b) mais do que 92,5 kg \
c) menos do que 64 kg \
d) 64 kg \
e) 64 kg ou menos

```python
# Questão 1 - dados
# Como a balança registra centésimos de kg, usamos a correção de 0,005 kg nos limites
n = 500
mu = 75.50
sigma = 7.50

# a) entre 60 e 77,5 kg  ->  59,995 < X < 77,505
p_a = norm.cdf(77.505, loc=mu, scale=sigma) - norm.cdf(59.995, loc=mu, scale=sigma)
print(f'a) P = {p_a:.4f} -> {p_a * n:.0f} estudantes')
```

```python
# b) mais do que 92,5 kg  ->  X > 92,505
p_b = norm.sf(92.505, loc=mu, scale=sigma)
print(f'b) P = {p_b:.4f} -> {p_b * n:.0f} estudantes')
```

```python
# c) menos do que 64 kg  ->  X < 63,995
p_c = norm.cdf(63.995, loc=mu, scale=sigma)
print(f'c) P = {p_c:.4f} -> {p_c * n:.0f} estudantes')
```

```python
# d) exatamente 64 kg  ->  63,995 < X < 64,005
p_d = norm.cdf(64.005, loc=mu, scale=sigma) - norm.cdf(63.995, loc=mu, scale=sigma)
print(f'd) P = {p_d:.6f} -> {p_d * n:.2f} estudantes (praticamente nenhum)')
```

```python
# e) 64 kg ou menos  ->  X < 64,005
p_e = norm.cdf(64.005, loc=mu, scale=sigma)
print(f'e) P = {p_e:.4f} -> {p_e * n:.0f} estudantes')
```

Questão 2 \
Aparelhos de TVs são substituídos normalmente com uma média de 8,2 anos e desvio padrão de 1,1 ano.
a) Determine a probabilidade de que uma TV deva ser substituída em menos de 5,0 anos. \
b) Se você deseja estabelecer uma garantia de modo que apenas 1% dos aparelhos seja substituídos antes da expiração da garantia, qual deve ser o prazo ? Dica: use a função ppf

```python
# Questão 2
mu = 8.2
sigma = 1.1

# a) P(X < 5,0 anos)
p = norm.cdf(5.0, loc=mu, scale=sigma)
print(f'a) P(X < 5) = {p:.6f} ({p * 100:.4f}%)')
```

```python
# b) prazo de garantia t tal que P(X < t) = 0,01
t = norm.ppf(0.01, loc=mu, scale=sigma)
print(f'b) Prazo de garantia = {t:.2f} anos')
```

Questão 3 \
A média dos diâmetros internos de uma amostra de 200 arruelas produzidas por certa máquina é 0,502” e o desvio padrão 0,005”. A finalidade para a qual essas arruelas são produzidas permite a tolerância máxima de 0,496 a 0,508 polegadas de diâmetro. Determine a porcentagem de arruelas defeituosas produzidas, admitindo-se que os diâmetros são normalmente distribuídos.

```python
# Questão 3
mu = 0.502
sigma = 0.005

# Arruelas aceitas: 0,496 <= X <= 0,508
p_ok = norm.cdf(0.508, loc=mu, scale=sigma) - norm.cdf(0.496, loc=mu, scale=sigma)
p_def = 1 - p_ok
print(f'Porcentagem de arruelas defeituosas: {p_def * 100:.2f}%')
print(f'Em 200 arruelas: aproximadamente {p_def * 200:.0f} defeituosas')
```

Questão 4 \
Suponha que as temperaturas do corpo humano sejam normalmente distribuídas, com média 36,8°C e desvio padrão 0,35°C. \
a) Um hospital usa 38,1°C como a menor temperatura considerada como febre. Qual é porcentagem de pessoas normais e saudáveis que seriam consideradas febris? Tal porcentagem sugere que o corte em 38,1°C seja apropriado? \
b) Os médicos desejam selecionar uma temperatura mínima para requerimento de exames adicionais. Qual deve ser essa
temperatura, se eles desejam que somente 5% das pessoas saudáveis tenham temperatura superior ? (Tal resultado é chamado falso positivo. Teste positivo, mas a pessoa não está doente).

```python
# Questão 4
mu = 36.8
sigma = 0.35

# a) P(X > 38,1)
p = norm.sf(38.1, loc=mu, scale=sigma)
print(f'a) Porcentagem considerada febril: {p * 100:.4f}%')
print('Como a porcentagem é muito pequena, o corte em 38,1 °C é apropriado:')
print('quase nenhuma pessoa saudável seria classificada como febril.')
```

```python
# b) temperatura t tal que P(X > t) = 0,05  ->  P(X < t) = 0,95
t = norm.ppf(0.95, loc=mu, scale=sigma)
print(f'b) Temperatura mínima para exames adicionais: {t:.2f} °C')
```
