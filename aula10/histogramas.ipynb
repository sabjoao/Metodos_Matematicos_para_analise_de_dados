## Criação de histogramas com Seaborn
Nesta aula analisaremos o índice de percepção de corrupção no setor público nos diferentes países do globo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('CPI_2014.csv', encoding='latin-1')
```

```python
df.head()
```

```python
sns.displot(df['CPI_2014'], kde = False)
sns.set_style('darkgrid')
plt.title('Indice de percepcao de corrupcao - 2014')
plt.xlabel('Nivel de corrupcao')
plt.axvline(60, color = 'k', linestyle = '--') #coloca uma linha no axis vertical, onde o valor for 60
plt.show()
```

Exercício \
Criar um histograma apenas com países da Europa.

```python
df2 = df[df['Continent']=='Europe']
df2.head(5)
```

```python
sns.displot(df2['CPI_2014'], kde = False, bins = 15) #bins mostra a quantidade de colunas
sns.set_style('darkgrid')
plt.title('Indice de percepcao de corrupcao - 2014')
plt.xlabel('Nivel de corrupcao')
plt.axvline(60, color = 'k', linestyle = '--') #coloca uma linha no axis vertical, onde o valor for 60
plt.show()
```
